###Matrix multiplication

Two main ways
1. Element _ wise implementation
2. Matix multiplication (dot Product "." -> a.b)

```python
# Element - wise
tensor = torch.tensor([2,3,4])
tensor * tensor
```

```python
#Matrix Multiplication
torch.matmul(tensor,tensor)
```

In matrix multiplication do remember the how matrix multiplication is done in maths with poper dimension rule and result dimension.
In pytorch "matmul( m1, m2)" for matrix multiplication ofm1,m1 also "mm(m1,m2)" aa mm is alias of matmul

###Transpose of Matrix

```python
tensor_1 = torch.tensor([[3,4,5],
                         [1,5,8]])
tensor_1
```

```python
tensor_2 = tensor_1.T
# switched the dimension by transpose as same as maths

tensor_3 = torch.matmul(tensor_1,tensor_2)
```
