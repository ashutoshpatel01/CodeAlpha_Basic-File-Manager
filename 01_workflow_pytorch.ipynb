#Pytorch Work Flows

#### "nn" contains all bulding blocks for Neural network

```python
import torch
from torch import nn
import matplotlib.pyplot as plt

torch.__version__
```

##1. Data preparing and loading
exel, images, videos,DNA, text

ML 2 parts
1. Get data into numerical
2. Build a model to learn patterns

```python
#1. Creating the known parameters'
weight = 0.7
bias = 0.3

# Creata
start = 0
end = 1
step = 0.02
X = torch.arange(start, end, step).unsqueeze(dim=1)
y = weight * X + bias

X[:10], y[:10]
```

```python
len(X), len(y)
```

##Splitting data in test and train

```python
# Creat test train split

train_split = int (0.8 * len(X))
X_train, y_train = X[:train_split], X[: train_split]

X_test, y_test = X[: train_split:],y[train_split:]

len(X_train), len(y_train), len(X_test), len(y_test)
```

####Visualizing data

```python

def plot_prediction(train_data = X_train,
                    train_labels = y_train,
                    test_data = X_test,
                    test_labels = y_test,
                    plot_prediction = None):
  plt.figure(figsize = (4 , 4))
  plt.scatter(train_data, train_labels, c = "b", s = 4, label = "Training Data")
  plt.scatter(test_data, test_labels, c = "g", s = 4, label = "Test Data")
  if plot_prediction is not None:
    plt.plot(test_data, plot_prediction, c = "r", label = "Prediction")
  plt.legend()
  plt.show()
```

```python
plot_prediction()
```

#### bUILD model

```python
#Creting the linear regresion model
class LinearRegression(nn.Module):
  def __init__(self):
    super().__init__()
    self.weights = nn.Parameter(torch.randn(1,
                                            requires_grad = True,
                                            dtype = torch.float))
    self.bias = nn.Parameter(torch.randn(1,
                                         requires_grad = True,
                                         dtype = torch.float))
    #forward method  to define the computation in the model
    def forward(self, x: torch.Tensor) -> torch.Tensor:
      return self.weights * x + self.bias
```
